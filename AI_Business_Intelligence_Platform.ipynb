# 🤖 Machine Learning Preprocessing

```python
# ================================
# Machine Learning Libraries
# ================================

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import LabelEncoder
from sklearn.metrics import (
    accuracy_score,
    classification_report,
    confusion_matrix,
    ConfusionMatrixDisplay
)

from sklearn.linear_model import LogisticRegression
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier

import joblib
```

## Machine Learning Data Preparation

Create an independent modeling dataframe, remove identifiers and constant columns, repair missing values, encode categorical features, and split features from the Attrition target.


```python
# Import the preprocessing tools needed for a reusable model pipeline.
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder
from sklearn.pipeline import Pipeline
from sklearn.metrics import precision_score, recall_score, f1_score

# Build a fresh modeling copy and remove columns that are identifiers or constant.
ml_df = df.copy()
ml_df = ml_df.drop(columns=["EmployeeCount", "EmployeeNumber", "Over18", "StandardHours"], errors="ignore")

# Convert the literal Nan token to a real missing value and inspect remaining missingness.
ml_df = ml_df.replace(r"^\s*Nan\s*$", np.nan, regex=True)
missing_before = ml_df.isna().sum()
print("Missing values before handling:")
display(missing_before[missing_before > 0])

# Keep rows with a known target; fill feature gaps using the mode or median, as appropriate.
ml_df = ml_df.dropna(subset=["Attrition"])
categorical_columns = ml_df.drop(columns="Attrition").select_dtypes(include=["object", "category"]).columns.tolist()
numeric_columns = ml_df.drop(columns="Attrition").select_dtypes(include=[np.number]).columns.tolist()
for column in categorical_columns:
    if ml_df[column].isna().any():
        ml_df[column] = ml_df[column].fillna(ml_df[column].mode().iloc[0])
for column in numeric_columns:
    if ml_df[column].isna().any():
        ml_df[column] = ml_df[column].fillna(ml_df[column].median())
print("Remaining missing values:", int(ml_df.isna().sum().sum()))
print("Categorical columns to one-hot encode:", categorical_columns)
print("Numeric/ordinal columns passed through as numbers:", numeric_columns)

# Separate Attrition as the target and use a stratified 80/20 split with the requested seed.
X = ml_df.drop(columns="Attrition")
y = ml_df["Attrition"]
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.20, random_state=42, stratify=y
)

# Encode nominal categories; leave integer and ordinal columns as numeric values.
preprocessor = ColumnTransformer(
    [("categorical", OneHotEncoder(handle_unknown="ignore"), categorical_columns)],
    remainder="passthrough",
    verbose_feature_names_out=False,
)
```

## Train and Evaluate Models

Train Logistic Regression, Decision Tree, and Random Forest on the same training split. Compare their test metrics and confusion matrices; select the model with the highest F1-score for Attrition="Yes".


```python
# Refit the previous setup and the class-balanced models on the same split to compare their test metrics.
from sklearn.metrics import roc_auc_score, roc_curve

# These settings reproduce the prior run: LR/DT unweighted and RF already balanced.
previous_estimators = {
    "Logistic Regression": LogisticRegression(max_iter=2000, random_state=42),
    "Decision Tree": DecisionTreeClassifier(random_state=42),
    "Random Forest": RandomForestClassifier(n_estimators=300, random_state=42, class_weight="balanced"),
}
# Apply balanced class weights to every requested model.
balanced_estimators = {
    "Logistic Regression": LogisticRegression(max_iter=2000, random_state=42, class_weight="balanced"),
    "Decision Tree": DecisionTreeClassifier(random_state=42, class_weight="balanced"),
    "Random Forest": RandomForestClassifier(n_estimators=300, random_state=42, class_weight="balanced"),
}

def evaluate_model_set(estimators, show_confusion_matrices=False):
    fitted_models = {}
    metric_rows = []
    roc_curves = {}
    for model_name, estimator in estimators.items():
        pipeline = Pipeline([("preprocessor", preprocessor), ("model", estimator)])
        pipeline.fit(X_train, y_train)
        predictions = pipeline.predict(X_test)
        yes_index = list(pipeline.classes_).index("Yes")
        yes_scores = pipeline.predict_proba(X_test)[:, yes_index]
        fpr, tpr, _ = roc_curve(y_test, yes_scores, pos_label="Yes")
        auc_value = roc_auc_score((y_test == "Yes").astype(int), yes_scores)
        fitted_models[model_name] = pipeline
        roc_curves[model_name] = (fpr, tpr, auc_value)
        metric_rows.append({
            "Model": model_name,
            "Accuracy": accuracy_score(y_test, predictions),
            "Precision": precision_score(y_test, predictions, pos_label="Yes", zero_division=0),
            "Recall": recall_score(y_test, predictions, pos_label="Yes", zero_division=0),
            "F1-score": f1_score(y_test, predictions, pos_label="Yes", zero_division=0),
            "ROC-AUC": auc_value,
        })
        if show_confusion_matrices:
            print(f"{model_name} balanced model confusion matrix (rows/columns: No, Yes):")
            display(pd.DataFrame(
                confusion_matrix(y_test, predictions, labels=["No", "Yes"]),
                index=["Actual No", "Actual Yes"],
                columns=["Predicted No", "Predicted Yes"],
            ))
    return fitted_models, pd.DataFrame(metric_rows), roc_curves

# Train both model sets against the same prepared train/test split.
previous_models, previous_results, previous_roc = evaluate_model_set(previous_estimators)
trained_models, results_df, balanced_roc = evaluate_model_set(
    balanced_estimators, show_confusion_matrices=True
)

# Compare each balanced metric with its previous-run counterpart (change = balanced minus previous).
metric_names = ["Accuracy", "Precision", "Recall", "F1-score", "ROC-AUC"]
comparison_df = previous_results.merge(
    results_df, on="Model", suffixes=("_Previous", "_Balanced")
)
for metric in metric_names:
    comparison_df[f"{metric}_Change"] = (
        comparison_df[f"{metric}_Balanced"] - comparison_df[f"{metric}_Previous"]
    )
print("Previous vs class-balanced test metrics; positive change means the metric increased:")
display(comparison_df[["Model"] + [
    column
    for metric in metric_names
    for column in (f"{metric}_Previous", f"{metric}_Balanced", f"{metric}_Change")
]])

# Plot ROC curves for both versions to compare their ability to rank attrition cases.
plt.figure(figsize=(9, 7))
for model_name, (fpr, tpr, auc_value) in previous_roc.items():
    plt.plot(fpr, tpr, linestyle="--", alpha=0.75,
             label=f"Previous {model_name} (AUC={auc_value:.3f})")
for model_name, (fpr, tpr, auc_value) in balanced_roc.items():
    plt.plot(fpr, tpr, linewidth=2,
             label=f"Balanced {model_name} (AUC={auc_value:.3f})")
plt.plot([0, 1], [0, 1], linestyle=":", color="gray", label="Chance")
plt.xlabel("False Positive Rate")
plt.ylabel("True Positive Rate")
plt.title("ROC Curves: Previous vs Class-Balanced Models")
plt.legend(loc="lower right", fontsize=8)
plt.grid(alpha=0.25)
plt.tight_layout()
plt.show()

# Select the balanced model with the strongest Attrition='Yes' F1-score.
results_df = results_df.sort_values(
    ["F1-score", "ROC-AUC", "Recall", "Accuracy"], ascending=False
).reset_index(drop=True)
print("Class-balanced model ranking:")
display(results_df)
best_model_name = results_df.loc[0, "Model"]
best_model = trained_models[best_model_name]
print(f"Best class-balanced model by Attrition='Yes' F1-score: {best_model_name}")
```

## ROC-AUC Evaluation

ROC-AUC (area under the Receiver Operating Characteristic curve) measures how well a model ranks positive cases above negative cases across classification thresholds. The ROC curve plots the true positive rate against the false positive rate. An AUC of 0.5 is chance-level; 1.0 is perfect ranking. It complements threshold-based metrics such as accuracy, precision, recall, and F1-score.

This section compares the trained models on the held-out test set.

```python
# Calculate ROC-AUC for each trained model version, then compare them on one plot.
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.metrics import roc_auc_score, roc_curve

# Include each model set available in this notebook's current pipeline.
roc_model_groups = {}
for group_name, variable_name in [
    ("Previous", "previous_models"),
    ("Balanced", "trained_models"),
    ("Tuned", "tuned_models"),
]:
    model_group = globals().get(variable_name)
    if isinstance(model_group, dict) and model_group:
        roc_model_groups[group_name] = model_group

if not roc_model_groups:
    raise NameError("No fitted model dictionaries found. Run the model training cells first.")

roc_auc_rows = []
roc_curve_data = {}
for version_name, model_group in roc_model_groups.items():
    for model_name, fitted_model in model_group.items():
        class_labels = list(fitted_model.classes_)
        positive_index = class_labels.index("Yes")
        if hasattr(fitted_model, "predict_proba"):
            positive_scores = fitted_model.predict_proba(X_test)[:, positive_index]
        else:
            positive_scores = fitted_model.decision_function(X_test)
        y_binary = (y_test == "Yes").astype(int)
        fpr, tpr, _ = roc_curve(y_binary, positive_scores)
        auc_value = roc_auc_score(y_binary, positive_scores)
        display_name = f"{version_name} {model_name}"
        roc_curve_data[display_name] = (fpr, tpr, auc_value)
        roc_auc_rows.append({
            "Model Version": version_name,
            "Model": model_name,
            "ROC-AUC": auc_value,
        })

roc_auc_results = (
    pd.DataFrame(roc_auc_rows)
    .sort_values("ROC-AUC", ascending=False)
    .reset_index(drop=True)
)
display(roc_auc_results)

best_auc_row = roc_auc_results.iloc[0]
print(
    f"Best ROC-AUC: {best_auc_row['Model Version']} "
    f"{best_auc_row['Model']} (AUC={best_auc_row['ROC-AUC']:.4f})."
)

plt.figure(figsize=(11, 8))
for display_name, (fpr, tpr, auc_value) in roc_curve_data.items():
    plt.plot(fpr, tpr, linewidth=1.8, label=f"{display_name} (AUC={auc_value:.3f})")
plt.plot([0, 1], [0, 1], linestyle="--", color="gray", label="Chance (AUC=0.5)")
plt.xlabel("False Positive Rate")
plt.ylabel("True Positive Rate")
plt.title("ROC Curves for Trained Attrition Models")
plt.legend(loc="lower right", fontsize=8)
plt.grid(alpha=0.25)
plt.tight_layout()
plt.show()
```

## Precision Recall Analysis

A Precision-Recall (PR) curve shows the tradeoff between precision (the share of predicted attrition cases that truly leave) and recall (the share of actual attrition cases the model finds). It focuses on the positive class, which is useful here because attrition is much less common than retention. ROC curves can look strong even when a model misses many of the rare positive cases.

For proactive retention outreach, missed leavers can be costly, so recall matters. If the organization can offer low-cost support to many employees, I recommend the class-balanced Logistic Regression model as the initial screening model: it has the strongest Attrition="Yes" F1-score in the current comparison and substantially higher recall than the unweighted baseline. Review its precision and available outreach capacity before deployment, and tune the decision threshold to balance intervention cost against missed attrition.

```python
# Plot precision-recall curves and average precision for each fitted model version.
from sklearn.metrics import precision_recall_curve, average_precision_score

pr_model_groups = {}
for group_name, variable_name in [
    ("Previous", "previous_models"),
    ("Balanced", "trained_models"),
    ("Tuned", "tuned_models"),
]:
    model_group = globals().get(variable_name)
    if isinstance(model_group, dict) and model_group:
        pr_model_groups[group_name] = model_group

pr_rows = []
pr_curve_data = {}
y_test_binary = (y_test == "Yes").astype(int)

for version_name, model_group in pr_model_groups.items():
    for model_name, fitted_model in model_group.items():
        yes_index = list(fitted_model.classes_).index("Yes")
        positive_scores = fitted_model.predict_proba(X_test)[:, yes_index]
        precision_values, recall_values, _ = precision_recall_curve(
            y_test_binary, positive_scores
        )
        avg_precision = average_precision_score(y_test_binary, positive_scores)
        display_name = f"{version_name} {model_name}"
        pr_curve_data[display_name] = (
            precision_values, recall_values, avg_precision
        )
        pr_rows.append({
            "Model Version": version_name,
            "Model": model_name,
            "Average Precision": avg_precision,
        })

pr_results = (
    pd.DataFrame(pr_rows)
    .sort_values("Average Precision", ascending=False)
    .reset_index(drop=True)
)
display(pr_results)

plt.figure(figsize=(10, 7))
for display_name, (precision_values, recall_values, avg_precision) in pr_curve_data.items():
    plt.plot(
        recall_values, precision_values,
        linewidth=1.8,
        label=f"{display_name} (AP={avg_precision:.3f})",
    )
plt.xlabel("Recall")
plt.ylabel("Precision")
plt.title("Precision-Recall Curves for Trained Attrition Models")
plt.legend(loc="best", fontsize=8)
plt.grid(alpha=0.25)
plt.tight_layout()
plt.show()
```

## Cross Validation

Stratified five-fold cross-validation evaluates the selected final model on five different, class-balanced partitions of the training data. Using several validation folds shows how much results vary across splits and gives a steadier estimate than relying on one train-test split. The held-out test set remains separate for final evaluation.

```python
from sklearn.model_selection import StratifiedKFold, cross_validate
from sklearn.metrics import make_scorer, precision_score, recall_score, f1_score

# Score the selected pipeline on five stratified folds of the training set.
cv = StratifiedKFold(n_splits=5, shuffle=True, random_state=42)
scoring = {
    "accuracy": "accuracy",
    "precision": make_scorer(precision_score, pos_label="Yes", zero_division=0),
    "recall": make_scorer(recall_score, pos_label="Yes", zero_division=0),
    "f1": make_scorer(f1_score, pos_label="Yes", zero_division=0),
}
cv_scores = cross_validate(best_model, X_train, y_train, cv=cv, scoring=scoring, n_jobs=-1)
cv_summary = pd.DataFrame([
    {"Metric": metric.title(), "Mean": cv_scores[f"test_{metric}"].mean(), "Standard Deviation": cv_scores[f"test_{metric}"].std()}
    for metric in scoring
])
display(cv_summary)
```

## Model Predictions

This section samples five employees from the held-out test set and reports the actual outcome, predicted outcome, and predicted probability of Attrition = Yes. Any mismatches are labeled as missed leavers (false negatives) or unnecessary outreach flags (false positives) so they can be reviewed in context.

```python
# Sample five held-out employees and show actual outcomes, predictions, and attrition probabilities.
prediction_indices = X_test.sample(n=min(5, len(X_test)), random_state=42).index
prediction_input = X_test.loc[prediction_indices]
predicted_labels = best_model.predict(prediction_input)
yes_column = list(best_model.classes_).index("Yes")
attrition_probabilities = best_model.predict_proba(prediction_input)[:, yes_column]
predictions_df = pd.DataFrame({
    "Test Row": prediction_indices,
    "Actual Attrition": y_test.loc[prediction_indices].values,
    "Predicted Attrition": predicted_labels,
    "Prediction Probability (Yes)": attrition_probabilities,
})
display(predictions_df.style.format({"Prediction Probability (Yes)": "{:.1%}"}))

incorrect_predictions = predictions_df[predictions_df["Actual Attrition"] != predictions_df["Predicted Attrition"]]
if incorrect_predictions.empty:
    print("No incorrect predictions in this five-employee sample.")
else:
    for _, row in incorrect_predictions.iterrows():
        if row["Actual Attrition"] == "Yes":
            print("Test row", row["Test Row"], ": false negative — actual leaver predicted to stay; this is a missed retention opportunity.")
        else:
            print("Test row", row["Test Row"], ": false positive — predicted to leave but stayed; outreach may be unnecessary.")
```

## Hyperparameter Tuning

Use RandomizedSearchCV with stratified five-fold cross-validation to tune class-balanced Logistic Regression, Decision Tree, and Random Forest pipelines. The search optimizes F1-score for Attrition = Yes. Compare tuned models with the existing class-balanced models on the same held-out test set, select the strongest tuned model by F1-score (ROC-AUC, recall, and accuracy break ties), and save it as attrition_model.pkl.

```python
# Tune each class-balanced pipeline using randomized search and stratified five-fold cross-validation.
from sklearn.model_selection import RandomizedSearchCV, StratifiedKFold
from sklearn.metrics import make_scorer, roc_auc_score

attrition_f1_scorer = make_scorer(f1_score, pos_label="Yes", zero_division=0)
tuning_cv = StratifiedKFold(n_splits=5, shuffle=True, random_state=42)

# Class weighting keeps the minority attrition class in view during tuning.
tuning_spaces = {
    "Logistic Regression": (
        LogisticRegression(max_iter=3000, random_state=42, class_weight="balanced"),
        {"model__C": [0.01, 0.1, 1, 10, 100], "model__penalty": ["l1", "l2"], "model__solver": ["liblinear"]},
    ),
    "Decision Tree": (
        DecisionTreeClassifier(random_state=42, class_weight="balanced"),
        {
            "model__max_depth": [None, 5, 10, 20, 30],
            "model__min_samples_split": [2, 5, 10, 20],
            "model__min_samples_leaf": [1, 2, 4, 8],
            "model__max_features": [None, "sqrt", "log2"],
        },
    ),
    "Random Forest": (
        RandomForestClassifier(random_state=42, class_weight="balanced"),
        {
            "model__n_estimators": [100, 200, 300, 500],
            "model__max_depth": [None, 10, 20, 30],
            "model__min_samples_split": [2, 5, 10],
            "model__min_samples_leaf": [1, 2, 4],
            "model__max_features": ["sqrt", "log2", None],
        },
    ),
}

tuned_models = {}
tuning_rows = []
best_parameter_rows = []
for model_name, (estimator, parameter_space) in tuning_spaces.items():
    model_pipeline = Pipeline([("preprocessor", preprocessor), ("model", estimator)])
    search = RandomizedSearchCV(
        estimator=model_pipeline,
        param_distributions=parameter_space,
        n_iter=12,
        scoring=attrition_f1_scorer,
        cv=tuning_cv,
        random_state=42,
        n_jobs=-1,
        refit=True,
        verbose=1,
    )
    search.fit(X_train, y_train)
    tuned_pipeline = search.best_estimator_
    tuned_models[model_name] = tuned_pipeline
    best_parameter_rows.append({
        "Model": model_name,
        "Best CV F1 (Attrition=Yes)": search.best_score_,
        "Best Hyperparameters": search.best_params_,
    })

    test_predictions = tuned_pipeline.predict(X_test)
    yes_column = list(tuned_pipeline.classes_).index("Yes")
    test_probabilities = tuned_pipeline.predict_proba(X_test)[:, yes_column]
    tuning_rows.append({
        "Model": model_name,
        "Accuracy": accuracy_score(y_test, test_predictions),
        "Precision": precision_score(y_test, test_predictions, pos_label="Yes", zero_division=0),
        "Recall": recall_score(y_test, test_predictions, pos_label="Yes", zero_division=0),
        "F1-score": f1_score(y_test, test_predictions, pos_label="Yes", zero_division=0),
        "ROC-AUC": roc_auc_score((y_test == "Yes").astype(int), test_probabilities),
    })

# Show each model's chosen hyperparameters and held-out metrics.
best_parameters_df = pd.DataFrame(best_parameter_rows)
tuned_results_df = pd.DataFrame(tuning_rows)
print("Best hyperparameters selected by stratified five-fold CV:")
display(best_parameters_df)
print("Tuned model metrics on the held-out test set:")
display(tuned_results_df.sort_values(["F1-score", "ROC-AUC", "Recall", "Accuracy"], ascending=False).reset_index(drop=True))

# Compare tuned metrics with the earlier class-balanced scores on the same test split.
baseline_results_for_tuning = results_df[["Model", "Accuracy", "Precision", "Recall", "F1-score", "ROC-AUC"]].copy()
tuning_comparison_df = baseline_results_for_tuning.merge(
    tuned_results_df, on="Model", suffixes=("_Balanced", "_Tuned")
)
for metric in ["Accuracy", "Precision", "Recall", "F1-score", "ROC-AUC"]:
    tuning_comparison_df[f"{metric}_Change"] = (
        tuning_comparison_df[f"{metric}_Tuned"] - tuning_comparison_df[f"{metric}_Balanced"]
    )
print("Class-balanced baseline vs tuned metrics (change = tuned minus baseline):")
display(tuning_comparison_df)

# Keep the tuned candidate with the strongest positive-class F1 and save its full pipeline.
tuned_results_df = tuned_results_df.sort_values(
    ["F1-score", "ROC-AUC", "Recall", "Accuracy"], ascending=False
).reset_index(drop=True)
best_model_name = tuned_results_df.loc[0, "Model"]
best_model = tuned_models[best_model_name]
joblib.dump(best_model, "attrition_model.pkl")
print(f"Selected tuned model: {best_model_name}")
print("Updated tuned model saved to attrition_model.pkl")
```
